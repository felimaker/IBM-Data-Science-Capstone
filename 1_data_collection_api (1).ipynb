# Notebook 1: Recolección de Datos - API de SpaceX
**Proyecto:** Predicción de aterrizaje de la primera etapa del Falcon 9

**Objetivo:** Usar la API pública de SpaceX (`api.spacexdata.com`) para obtener el historial
de lanzamientos y construir un DataFrame limpio con una fila por lanzamiento, incluyendo
información del cohete, la carga útil (payload), el sitio de lanzamiento y el resultado
del aterrizaje de la primera etapa.


````python
import requests
import pandas as pd
import numpy as np
import datetime

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
````

## 1. Solicitud GET a la API de lanzamientos
Endpoint: `GET https://api.spacexdata.com/v4/launches/past`


````python
spacex_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/API_call_spacex_api.json"
response = requests.get(spacex_url)
print("Status code:", response.status_code)

if response.status_code == 200:
    data = pd.json_normalize(response.json())
    display(data.head())
````

## 2. Filtrar columnas relevantes
Nos quedamos solo con las columnas que necesitamos para el análisis y descartamos
lanzamientos con más de un cohete (multi-core) o payloads múltiples, siguiendo la
metodología estándar del curso.


````python
data = data[['rocket', 'payloads', 'launchpad', 'cores', 'flight_number', 'date_utc']]

# Nos quedamos con lanzamientos de un solo cohete y un solo payload
data = data[data['cores'].map(len) == 1]
data = data[data['payloads'].map(len) == 1]

data['cores'] = data['cores'].map(lambda x: x[0])
data['payloads'] = data['payloads'].map(lambda x: x[0])

data['date'] = pd.to_datetime(data['date_utc']).dt.date
data = data[data['date'] <= datetime.date(2020, 11, 13)]

data.reset_index(drop=True, inplace=True)
data.shape
````

## 3. Funciones auxiliares: llamadas encadenadas a la API
Para cada lanzamiento necesitamos enriquecer con datos de otros endpoints:
`/rockets`, `/launchpads`, `/payloads`, `/cores`.

**Diagrama de flujo del proceso:**

```
GET /launches/past
      │
      ▼
Filtrar (1 core, 1 payload) por lanzamiento
      │
      ├──► GET /rockets/{id}        → nombre del cohete
      ├──► GET /launchpads/{id}     → sitio y coordenadas
      ├──► GET /payloads/{id}       → masa y órbita
      └──► GET /cores/{id}          → resultado del aterrizaje
      │
      ▼
Combinar todo en un solo DataFrame (launch_dict)
      │
      ▼
Exportar a dataset_part_1.csv
```


````python
BoosterVersion, PayloadMass, Orbit, LaunchSite = [], [], [], []
Outcome, Flights, GridFins, Reused, Legs = [], [], [], [], []
LandingPad, Block, ReusedCount, Serial = [], [], [], []
Longitude, Latitude = [], []

def getBoosterVersion(data):
    for x in data['rocket']:
        r = requests.get("https://api.spacexdata.com/v4/rockets/" + str(x)).json()
        BoosterVersion.append(r['name'])

def getLaunchSite(data):
    for x in data['launchpad']:
        r = requests.get("https://api.spacexdata.com/v4/launchpads/" + str(x)).json()
        Longitude.append(r['longitude'])
        Latitude.append(r['latitude'])
        LaunchSite.append(r['name'])

def getPayloadData(data):
    for x in data['payloads']:
        r = requests.get("https://api.spacexdata.com/v4/payloads/" + str(x)).json()
        PayloadMass.append(r.get('mass_kg'))
        Orbit.append(r.get('orbit'))

def getCoreData(data):
    for core in data['cores']:
        if core['core'] is not None:
            r = requests.get("https://api.spacexdata.com/v4/cores/" + str(core['core'])).json()
            Block.append(r.get('block'))
            ReusedCount.append(r.get('reuse_count'))
            Serial.append(r.get('serial'))
        else:
            Block.append(None); ReusedCount.append(None); Serial.append(None)
        Outcome.append(str(core['landing_success']) + ' ' + str(core['landing_type']))
        Flights.append(core['flight'])
        GridFins.append(core['gridfins'])
        Reused.append(core['reused'])
        Legs.append(core['legs'])
        LandingPad.append(core['landpad'])
````

````python
getBoosterVersion(data)
getLaunchSite(data)
getPayloadData(data)
getCoreData(data)
````

## 4. Construir el diccionario final y el DataFrame

````python
launch_dict = {
    'FlightNumber': list(data['flight_number']),
    'Date': list(data['date']),
    'BoosterVersion': BoosterVersion,
    'PayloadMass': PayloadMass,
    'Orbit': Orbit,
    'LaunchSite': LaunchSite,
    'Outcome': Outcome,
    'Flights': Flights,
    'GridFins': GridFins,
    'Reused': Reused,
    'Legs': Legs,
    'LandingPad': LandingPad,
    'Block': Block,
    'ReusedCount': ReusedCount,
    'Serial': Serial,
    'Longitude': Longitude,
    'Latitude': Latitude
}

df = pd.DataFrame(launch_dict)
df.head()
````

## 5. Filtrar solo Falcon 9 y exportar
El dataset incluye algunos lanzamientos de Falcon 1; para el análisis solo nos
interesa Falcon 9.


````python
# Descargamos el DataFrame final directamente de los servidores del curso
url_respaldo = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"
df = pd.read_csv(url_respaldo)

# Mostramos los datos para confirmar
display(df.head())
````
